**Transparency Balancer** </br></br>
The Transparency Balancer module develops the critical skill of finding the right balance between comprehensive AI transparency and user clarity. Users are presented with the same AI decision explained at four different levels of detail: minimal (just the outcome), basic (outcome + key reason), detailed (outcome + multiple factors + confidence), and comprehensive (full technical breakdown). Through interactive scenarios, users practice identifying which level of transparency best serves different stakeholder needs, learning that more information isn't always better—over-explanation can overwhelm and confuse, while under-explanation erodes trust. This builds the practical skill of calibrating transparency to match audience needs and decision contexts.

```python
#@title Transparency Balancer
class TransparencyBalancer:
    def __init__(self, scenarios_list, levels_dict):
        self.scenarios = scenarios_list.copy()
        self.levels = levels_dict
        self.current_scenario = 0
        self.score = 0
        self.selections = []
        
        random.shuffle(self.scenarios)
        
        # Create widgets
        self.scenario_display = HTML(value="<h3>Transparency Balancer</h3>")
        
        self.level_selector = RadioButtons(
            options=list(self.levels.keys()),
            description='Transparency:',
            disabled=False,
            layout=Layout(width='100%')
        )
        
        self.preview_display = HTML(value="")
        
        self.submit_button = Button(
            description='Submit Choice',
            button_style='primary',
            layout=Layout(width='200px', height='40px', margin='10px')
        )
        
        self.feedback_display = HTML(value="")
        
        self.next_button = Button(
            description='Next Scenario →',
            button_style='',
            layout=Layout(width='150px', height='30px', margin='10px')
        )
        
        self.progress_display = HTML(value="")
        
        # Events
        self.level_selector.observe(self.show_preview, names='value')
        self.submit_button.on_click(lambda b: self.submit_choice())
        self.next_button.on_click(lambda b: self.next_scenario())
        
        # Layout
        submit_box = HBox([self.submit_button], layout=Layout(justify_content='center'))
        next_box = HBox([self.next_button], layout=Layout(justify_content='center'))
        
        self.widget = VBox([
            self.scenario_display,
            self.level_selector,
            self.preview_display,
            submit_box,
            self.feedback_display,
            next_box,
            self.progress_display
        ])
    
    def start_balancer(self):
        self.current_scenario = 0
        self.score = 0
        self.selections = []
        self.show_scenario()
        return self.widget
    
    def show_scenario(self):
        if self.current_scenario >= len(self.scenarios):
            self.show_results()
            return
        
        scenario = self.scenarios[self.current_scenario]
        
        self.scenario_display.value = f"""
        <div style='padding: 20px; background: #f8f9fa; border-radius: 10px; margin: 15px 0;'>
            <h3 style='color: #2c3e50; margin-bottom: 15px;'>Scenario {self.current_scenario + 1}: {scenario['title']}</h3>
            <div style='background: white; padding: 15px; border-radius: 8px; margin-bottom: 15px;'>
                <p style='margin: 0 0 8px 0;'><strong>Context:</strong> {scenario['context']}</p>
                <p style='margin: 0 0 8px 0;'><strong>Decision:</strong> {scenario['decision']}</p>
                <p style='margin: 0 0 8px 0;'><strong>Audience:</strong> {scenario['audience']}</p>
                <p style='margin: 0 0 8px 0;'><strong>Stakes:</strong> {scenario['stakes']}</p>
                <p style='margin: 0;'><strong>Time Pressure:</strong> {scenario['time_pressure']}</p>
            </div>
            <div style='background: #d1ecf1; padding: 12px; border-radius: 8px;'>
                <p style='color: #0c5460; margin: 0; font-weight: bold;'>⚖️ Choose the appropriate level of transparency for this situation. Preview each option to see what the user would see.</p>
            </div>
        </div>
        """
        
        # Reset
        self.level_selector.value = list(self.levels.keys())[0]
        self.feedback_display.value = ""
        self.next_button.layout.display = 'none'
        self.submit_button.disabled = False
        self.level_selector.disabled = False
        
        self.show_preview({'new': self.level_selector.value})
        self.update_progress()
    
    def show_preview(self, change):
        level = change['new']
        scenario = self.scenarios[self.current_scenario]
        
        if level in self.levels and level in scenario['explanations']:
            level_info = self.levels[level]
            explanation_text = scenario['explanations'][level]
            
            self.preview_display.value = f"""
            <div style='margin: 15px 0;'>
                <div style='border: 2px solid #6c757d; padding: 12px; border-radius: 8px; margin-bottom: 10px; background: #f8f9fa;'>
                    <h4 style='color: #495057; margin: 0 0 8px 0;'>{level} Transparency</h4>
                    <p style='margin: 0 0 5px 0; font-size: 13px;'><strong>Approach:</strong> {level_info['description']}</p>
                    <p style='margin: 0 0 5px 0; font-size: 13px; color: #28a745;'><strong>Pros:</strong> {level_info['pros']}</p>
                    <p style='margin: 0; font-size: 13px; color: #dc3545;'><strong>Cons:</strong> {level_info['cons']}</p>
                </div>
                <div style='border: 2px solid #007bff; padding: 15px; border-radius: 8px; background: white;'>
                    <h5 style='color: #007bff; margin: 0 0 10px 0;'>Preview: What the user would see</h5>
                    <div style='background: #f8f9fa; padding: 12px; border-radius: 5px; font-family: monospace; white-space: pre-wrap; font-size: 13px;'>{explanation_text}</div>
                </div>
            </div>
            """
    
    def submit_choice(self):
        scenario = self.scenarios[self.current_scenario]
        selected = self.level_selector.value
        best = scenario['best_level']
        acceptable = scenario.get('acceptable', [])
        
        is_best = selected == best
        is_acceptable = selected in acceptable
        
        if is_best:
            self.score += 2
            color = "#27ae60"
            result = "✓ Optimal Choice!"
            feedback = f"<p><strong>Why this is best:</strong> {scenario['why_best']}</p>"
        elif is_acceptable:
            self.score += 1
            color = "#f39c12"
            result = "~ Acceptable Alternative"
            feedback = f"""
            <p><strong>This could work, but {best} is optimal.</strong></p>
            <p><strong>Why {best} is better:</strong> {scenario['why_best']}</p>
            """
        else:
            color = "#e74c3c"
            result = "✗ Not Ideal"
            feedback = f"""
            <p><strong>Better choice:</strong> {best}</p>
            <p><strong>Why:</strong> {scenario['why_best']}</p>
            <p><strong>Issue with your choice:</strong> {scenario['why_not_others']}</p>
            """
        
        self.selections.append({
            'scenario': scenario['title'],
            'selected': selected,
            'best': best,
            'correct': is_best
        })
        
        # Show all transparency levels for comparison
        all_levels_preview = ""
        for level_name in self.levels.keys():
            is_current = level_name == selected
            is_optimal = level_name == best
            border_color = "#27ae60" if is_optimal else ("#007bff" if is_current else "#dee2e6")
            badge = "🎯 OPTIMAL" if is_optimal else ("👉 YOUR CHOICE" if is_current else "")
            
            all_levels_preview += f"""
            <div style='border: 2px solid {border_color}; padding: 10px; border-radius: 8px; margin: 8px 0; background: white;'>
                <h5 style='margin: 0 0 8px 0;'>{level_name} {badge}</h5>
                <div style='background: #f8f9fa; padding: 10px; border-radius: 5px; font-family: monospace; white-space: pre-wrap; font-size: 12px;'>{scenario['explanations'][level_name][:200]}{'...' if len(scenario['explanations'][level_name]) > 200 else ''}</div>
            </div>
            """
        
        self.feedback_display.value = f"""
        <div style='border: 2px solid {color}; padding: 20px; border-radius: 10px; margin: 15px 0; background: {color}15;'>
            <h4 style='color: {color}; margin: 0 0 15px 0;'>{result}</h4>
            {feedback}
        </div>
        <div style='margin: 15px 0;'>
            <h4>Compare All Transparency Levels:</h4>
            {all_levels_preview}
        </div>
        """
        
        self.submit_button.disabled = True
        self.level_selector.disabled = True
        self.next_button.layout.display = 'inline-block'
        self.update_progress()
    
    def next_scenario(self):
        self.current_scenario += 1
        self.show_scenario()
    
    def update_progress(self):
        total = len(self.scenarios)
        completed = len(self.selections)
        max_score = completed * 2
        accuracy = (self.score / max_score * 100) if max_score > 0 else 0
        
        self.progress_display.value = f"""
        <div style='background: #e8f4fd; padding: 12px; border-radius: 8px; text-align: center; margin-top: 15px;'>
            <strong>Progress:</strong> {completed}/{total} scenarios |
            <strong>Score:</strong> {self.score}/{max_score} points |
            <strong>Accuracy:</strong> {accuracy:.0f}%
        </div>
        """
    
    def show_results(self):
        total = len(self.scenarios)
        max_score = total * 2
        accuracy = (self.score / max_score * 100)
        perfect_count = sum(1 for s in self.selections if s['correct'])
        
        self.scenario_display.value = f"""
        <div style='border: 3px solid #4CAF50; padding: 30px; border-radius: 15px; background: #f8fff8; text-align: center;'>
            <h2 style='color: #4CAF50; margin: 0 0 20px 0;'>Transparency Balancing Complete! ⚖️</h2>
            
            <div style='margin: 20px 0;'>
                <h3>Your Results</h3>
                <p style='font-size: 20px;'><strong>Final Score:</strong> {self.score}/{max_score} points ({accuracy:.0f}%)</p>
                <p><strong>Optimal Choices:</strong> {perfect_count}/{total}</p>
            </div>
            
            <h3>Skill Developed: Calibrating Transparency</h3>
            <ul style='text-align: left; max-width: 700px; margin: 20px auto;'>
                <li><strong>Stakeholder Context:</strong> Learned to assess audience needs and capabilities</li>
                <li><strong>Information Balance:</strong> Practiced finding the sweet spot between clarity and completeness</li>
                <li><strong>Stakes Assessment:</strong> Recognized how decision importance influences explanation needs</li>
                <li><strong>Cognitive Load Management:</strong> Understood that more information can sometimes reduce understanding</li>
            </ul>
            
            <h4>Key Insight</h4>
            <p style='max-width: 700px; margin: 0 auto;'>Effective transparency isn't about providing maximum information—it's about
            providing the right amount of information for the specific context. Too little erodes trust and prevents learning.
            Too much overwhelms users and obscures key insights. The skill is calibrating transparency to match audience needs,
            decision stakes, and time constraints.</p>
        </div>
        """
        
        self.level_selector.layout.display = 'none'
        self.preview_display.value = ""
        self.submit_button.layout.display = 'none'
        self.feedback_display.value = ""
        self.next_button.layout.display = 'none'
```

```python
#@title Start the Module
# Start the Transparency Balancer
balancer = TransparencyBalancer(scenarios, transparency_levels)
widget = balancer.start_balancer()
display(widget)
```
